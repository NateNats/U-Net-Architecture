```python
import cv2
import matplotlib.pyplot as plt
import os

img_path_before = "C:/Users/Cerdas05/Skripshot/U-Net-Architecture/dataset/training/images/ISIC_0000001.jpg"
img_before = cv2.cvtColor(cv2.imread(img_path_before), 
                           cv2.COLOR_BGR2RGB)
img_path_after = "C:/Users/Cerdas05/Skripshot/U-Net-Architecture/processed/1_resize/training/images/ISIC_0000001.jpg"
img_after = cv2.cvtColor(cv2.imread(img_path_after), cv2.COLOR_BGR2RGB)

mask_before = cv2.imread("C:/Users/Cerdas05/Skripshot/U-Net-Architecture/dataset/training/masks/ISIC_0000001_segmentation.png", cv2.COLOR_BGR2GRAY)
mask_after = cv2.imread("C:/Users/Cerdas05/Skripshot/U-Net-Architecture/processed/1_resize/training/masks/ISIC_0000001_segmentation.png", cv2.COLOR_BGR2GRAY)

fig, axes = plt.subplots(2, 2, figsize=(10, 8))
fig.suptitle('Perbandingan Citra dan Mask Sebelum & Sesudah Resize',
             fontweight='bold', fontsize=12)

axes[0, 0].imshow(img_before)
axes[0, 0].set_title(f'Citra Asli\n{img_before.shape[1]}×{img_before.shape[0]} piksel')
axes[0, 0].axis('off')

axes[0, 1].imshow(img_after)
axes[0, 1].set_title(f'Citra Setelah Resize\n{img_after.shape[1]}×{img_after.shape[0]} piksel')
axes[0, 1].axis('off')

axes[1, 0].imshow(mask_before, cmap='gray')
axes[1, 0].set_title(f'Mask Asli\n{mask_before.shape[1]}×{mask_before.shape[0]} piksel')
axes[1, 0].axis('off')

axes[1, 1].imshow(mask_after, cmap='gray')
axes[1, 1].set_title(f'Mask Setelah Resize\n{mask_after.shape[1]}×{mask_after.shape[0]} piksel')
axes[1, 1].axis('off')

plt.tight_layout()
plt.savefig('C:/Users/Cerdas05/Skripshot/U-Net-Architecture/output/perbandingan_resize.png', dpi=150, bbox_inches='tight')
plt.show()
```
